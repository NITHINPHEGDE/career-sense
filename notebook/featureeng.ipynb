```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("../data/raw/data_after_eda.csv")
```

```python
df.head(5)
```

```python
df=pd.get_dummies(df,columns=["Employment Status"],drop_first=True)
df.head()
```

```python
df.drop(columns=["Company Name"],inplace=True)
df.head(5)
```

```python
def map_job_category(title):
    title = title.lower()

    if "android" in title:
        return "android"
    elif "ios" in title:
        return "ios"
    elif "backend" in title:
        return "backend"
    elif "frontend" in title:
        return "frontend"
    elif "python" in title:
        return "python"
    elif "java" in title:
        return "java"
    elif "data" in title:
        return "data"
    elif "test" in title:
        return "testing"
    else:
        return "other"

df["job_category"] = df["Job Roles"].apply(map_job_category)
df.drop(columns=["Job Roles"], inplace=True)
```

```python
df.head(5)
```

```python
clm=["Location","company_size","job_category"]
df=pd.get_dummies(df,columns=clm,drop_first=True)
df.head(2)
```

```python
to_int=["Employment Status_Full Time",	"Employment Status_Intern",	"Employment Status_Trainee",	"Location_Chennai",	"Location_Hyderabad"	,"Location_Jaipur"	,"Location_Kerala"	,"Location_Kolkata",	"Location_Madhya Pradesh",	"Location_Mumbai"	,"Location_New Delhi"	,"Location_Pune"	,"company_size_Mid"	,"company_size_Small",	"job_category_backend"	,"job_category_data"	,"job_category_frontend"	,"job_category_ios",	"job_category_java"	,"job_category_other"	,"job_category_python"	,"job_category_testing"]
df[to_int]=df[to_int].astype(int)
df.head(2)
```
